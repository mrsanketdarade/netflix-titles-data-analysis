- Missing data detection

**Columns with missing values** include fields such as `director`, `cast`, `country`, `date_added`, `rating`, and `duration`.

## Missing Data Treatment

```python
for col in ['director', 'cast', 'country', 'rating', 'duration']:
    if df[col].notna().any():
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

## Date and Numeric Formatting

```python
df['date_added'] = pd.to_datetime(df['date_added'].astype('string').str.strip(), format='mixed', errors='coerce')
df['release_year'] = pd.to_numeric(df['release_year'], errors='coerce')
df['duration_num'] = pd.to_numeric(df['duration'].astype('string').str.extract(r'(\d+)')[0], errors='coerce')
df[['date_added', 'release_year', 'duration', 'duration_num']].head()
```

## Step 3: Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt\nimport seaborn as sns
```

### Q1. What is the distribution of Netflix release years?

```python
plt.figure()
plt.hist(df['release_year'].dropna(), bins=30)
plt.title("Distribution of Netflix Release Years")
plt.xlabel("Release Year")
plt.ylabel("Frequency")
plt.show()
```

### Analysis
- The histogram shows how the catalog is distributed across release years.
- More recent years may contain a larger share of the titles in this dataset.

### Conclusion
- Release-year distribution helps describe how recent the available Netflix catalog is.

### Q2. How many Movies and TV Shows are in the dataset?

```python
df['type'].value_counts().plot(kind='bar')
plt.title("Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

### Analysis
- The bar chart compares the number of movies with the number of TV shows.

### Conclusion
- This chart identifies which content type is more common in the dataset.

### Q3. How does movie duration vary by release year?

```python
movies = df[df['type'] == 'Movie'].copy()
movies['duration_num'] = pd.to_numeric(movies['duration'].astype('string').str.extract(r'(\d+)')[0], errors='coerce')
sns.scatterplot(x='release_year', y='duration_num', data=movies)
plt.title("Movie Duration vs Release Year")
plt.xlabel("Release Year")
plt.ylabel("Duration (Minutes)")
plt.show()
```

### Analysis
- Each point represents a movie.
- The plot compares movie release year with runtime in minutes.

### Conclusion
- The scatter plot helps inspect runtime patterns over time; it does not by itself establish causation.

### Q4. How do release years differ between Movies and TV Shows?

```python
sns.boxplot(x='type', y='release_year', data=df)
plt.title("Release Year vs Content Type")
plt.xlabel("Content Type")
plt.ylabel("Release Year")
plt.show()
```

### Q5. Which content ratings appear most frequently?

```python
df['rating'].value_counts().head(10).plot(kind='bar')
plt.title("Top Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

### Analysis
- The chart shows the ten most frequent content ratings.

### Conclusion
- Rating counts summarize how titles are classified in this dataset.

### Q6. What is the correlation between movie release year and duration?

```python
movie_corr = movies[['release_year', 'duration_num']].corr()
sns.heatmap(movie_corr, annot=True, vmin=-1, vmax=1)
plt.title("Movie Release Year and Duration Correlation")
plt.show()
```

### Analysis
- The heatmap displays the Pearson correlation between release year and movie duration, using rows with valid numeric values.

### Conclusion
- Correlation describes the strength and direction of a linear relationship; it does not prove causation.
